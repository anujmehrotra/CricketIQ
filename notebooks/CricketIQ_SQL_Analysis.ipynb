# CricketIQ SQL Analysis 

## Module 0: Data Cleaning and Preparation 

```python
# Create cleaned DuckDB tables in one step

con.execute("""
CREATE OR REPLACE TABLE matches_clean AS

SELECT
    * EXCLUDE (
        season,
        date,
        team1,
        team2,
        toss_winner,
        winner,
        city,
        venue
    ),

    /* Standardize season values */
    CASE
        WHEN CAST(season AS VARCHAR) = '2007/08' THEN '2008'
        WHEN CAST(season AS VARCHAR) = '2009/10' THEN '2010'
        WHEN CAST(season AS VARCHAR) = '2020/21' THEN '2020'
        ELSE CAST(season AS VARCHAR)
    END AS season,

    /* Convert date to proper DATE datatype */
    CAST(date AS DATE) AS date,

    /* Standardize Team 1 */
    CASE
        WHEN team1 = 'Delhi Daredevils'
            THEN 'Delhi Capitals'
        WHEN team1 = 'Kings XI Punjab'
            THEN 'Punjab Kings'
        WHEN team1 = 'Royal Challengers Bangalore'
            THEN 'Royal Challengers Bengaluru'
        WHEN team1 = 'Rising Pune Supergiant'
            THEN 'Rising Pune Supergiants'
        ELSE team1
    END AS team1,

    /* Standardize Team 2 */
    CASE
        WHEN team2 = 'Delhi Daredevils'
            THEN 'Delhi Capitals'
        WHEN team2 = 'Kings XI Punjab'
            THEN 'Punjab Kings'
        WHEN team2 = 'Royal Challengers Bangalore'
            THEN 'Royal Challengers Bengaluru'
        WHEN team2 = 'Rising Pune Supergiant'
            THEN 'Rising Pune Supergiants'
        ELSE team2
    END AS team2,

    /* Standardize toss winner */
    CASE
        WHEN toss_winner = 'Delhi Daredevils'
            THEN 'Delhi Capitals'
        WHEN toss_winner = 'Kings XI Punjab'
            THEN 'Punjab Kings'
        WHEN toss_winner = 'Royal Challengers Bangalore'
            THEN 'Royal Challengers Bengaluru'
        WHEN toss_winner = 'Rising Pune Supergiant'
            THEN 'Rising Pune Supergiants'
        ELSE toss_winner
    END AS toss_winner,

    /* Standardize match winner */
    CASE
        WHEN winner = 'Delhi Daredevils'
            THEN 'Delhi Capitals'
        WHEN winner = 'Kings XI Punjab'
            THEN 'Punjab Kings'
        WHEN winner = 'Royal Challengers Bangalore'
            THEN 'Royal Challengers Bengaluru'
        WHEN winner = 'Rising Pune Supergiant'
            THEN 'Rising Pune Supergiants'
        ELSE winner
    END AS winner,

    /* Standardize city names */
    CASE
        WHEN city = 'Bangalore'
            THEN 'Bengaluru'
        ELSE city
    END AS city,

    /* Standardize venue names */
    CASE
        WHEN venue = 'Subrata Roy Sahara Stadium'
            THEN 'Maharashtra Cricket Association Stadium, Pune'

        WHEN venue = 'Feroz Shah Kotla'
            THEN 'Arun Jaitley Stadium, Delhi'

        WHEN venue = 'M Chinnaswamy Stadium'
            THEN 'M Chinnaswamy Stadium, Bengaluru'

        WHEN venue = 'Rajiv Gandhi International Stadium, Uppal'
            THEN 'Rajiv Gandhi International Stadium, Uppal, Hyderabad'

        WHEN venue = 'Eden Gardens, Kolkata'
            THEN 'Eden Gardens'

        WHEN venue = 'Punjab Cricket Association Stadium, Mohali'
            THEN 'Punjab Cricket Association IS Bindra Stadium'

        WHEN venue = 'Punjab Cricket Association IS Bindra Stadium, Mohali, Chandigarh'
            THEN 'Punjab Cricket Association IS Bindra Stadium'

        WHEN venue = 'Brabourne Stadium, Mumbai'
            THEN 'Brabourne Stadium'

        WHEN venue = 'Wankhede Stadium, Mumbai'
            THEN 'Wankhede Stadium'
        WHEN venue = 'MA Chidambaram Stadium, Chepauk'
             THEN 'MA Chidambaram Stadium'

        ELSE venue
    END AS venue

FROM matches;
""")

con.execute("""
CREATE OR REPLACE TABLE deliveries_clean AS

SELECT
    * EXCLUDE (
        batting_team,
        bowling_team
    ),

    /* Standardize batting team */
    CASE
        WHEN batting_team = 'Delhi Daredevils'
            THEN 'Delhi Capitals'
        WHEN batting_team = 'Kings XI Punjab'
            THEN 'Punjab Kings'
        WHEN batting_team = 'Royal Challengers Bangalore'
            THEN 'Royal Challengers Bengaluru'
        WHEN batting_team = 'Rising Pune Supergiant'
            THEN 'Rising Pune Supergiants'
        ELSE batting_team
    END AS batting_team,

    /* Standardize bowling team */
    CASE
        WHEN bowling_team = 'Delhi Daredevils'
            THEN 'Delhi Capitals'
        WHEN bowling_team = 'Kings XI Punjab'
            THEN 'Punjab Kings'
        WHEN bowling_team = 'Royal Challengers Bangalore'
            THEN 'Royal Challengers Bengaluru'
        WHEN bowling_team = 'Rising Pune Supergiant'
            THEN 'Rising Pune Supergiants'
        ELSE bowling_team
    END AS bowling_team

FROM deliveries;
""")

print("Data cleaning completed successfully.")
```

### Query 8: Team Win Percentage

#### Business Problem

Compare teams based on their overall match win percentage.

```python
con.execute("""
WITH matches_played AS (

    SELECT team1 AS team
    FROM matches_clean

    UNION ALL

    SELECT team2 AS team
    FROM matches_clean
),

team_matches AS (

    SELECT
        team,
        COUNT(*) AS matches_played
    FROM matches_played
    GROUP BY team
),

team_wins AS (

    SELECT
        winner AS team,
        COUNT(*) AS matches_won
    FROM matches_clean
    WHERE winner IS NOT NULL
    GROUP BY winner
)

SELECT
    tm.team,
    tm.matches_played,
    COALESCE(tw.matches_won, 0) AS matches_won,

    ROUND(
        100.0 * COALESCE(tw.matches_won, 0)
        / tm.matches_played,
        2
    ) AS win_percentage

FROM team_matches tm

LEFT JOIN team_wins tw
    ON tm.team = tw.team

ORDER BY win_percentage DESC;
""").df()
```

### Observation 

Gujarat Titans have the highest win percentage (62.22%) among all IPL franchises, despite playing only 45 matches. Among long-established teams, Chennai Super Kings lead with a remarkable 57.98% win rate, followed by Mumbai Indians (55.17%) and Kolkata Knight Riders (52.19%). Pune Warriors have the lowest win percentage at 26.09%.

### Query 10: Toss Win Converted into Match Win

### Business Problem

Measure how frequently each team won the match after winning the toss.

```python
con.execute("""
SELECT
    toss_winner AS team,
    COUNT(*) AS tosses_won,

    SUM(
        CASE
            WHEN toss_winner = winner THEN 1
            ELSE 0
        END
    ) AS matches_won_after_toss,

    ROUND(
        100.0 *
        SUM(
            CASE
                WHEN toss_winner = winner THEN 1
                ELSE 0
            END
        )
        / COUNT(*),
        2
    ) AS toss_conversion_percentage

FROM matches_clean

WHERE toss_winner IS NOT NULL

GROUP BY toss_winner 
ORDER BY toss_conversion_percentage DESC;
""").df()
```

### Observations

Among long-established franchises, Chennai Super Kings performed best, winning 75 of the 122 matches in which they won the toss, for a conversion rate of 61.48%. Pune Warriors had the lowest conversion rate at just 15%.

### Query 12: Batters with the Highest Batting Average

#### Business Problem

Identify the batters with the highest batting average, considering only players with a meaningful number of dismissals.

```python
con.execute("""
WITH batter_runs AS (
    SELECT
        batter,
        CAST(SUM(batsman_runs) AS INTEGER) AS total_runs
    FROM deliveries_clean
    GROUP BY batter
),

batter_dismissals AS (
    SELECT
        player_dismissed AS batter,
        COUNT(*) AS dismissals
    FROM deliveries_clean
    WHERE player_dismissed IS NOT NULL
      AND dismissal_kind NOT IN (
          'retired hurt',
          'retired out',
          'obstructing the field'
      )
    GROUP BY player_dismissed
)

SELECT
    br.batter,
    br.total_runs,
    COALESCE(bd.dismissals, 0) AS dismissals,
    ROUND(
        br.total_runs * 1.0 / NULLIF(bd.dismissals, 0),
        2
    ) AS batting_average
FROM batter_runs br
JOIN batter_dismissals bd
    ON br.batter = bd.batter
WHERE bd.dismissals >= 20
ORDER BY batting_average DESC, br.batter
LIMIT 10;
""").df()
```

### Observation 

B Sai Sudharsan has the highest batting average (49.24) among batters with at least 20 dismissals, followed by KL Rahul (44.66) and Ruturaj Gaikwad (41.75). Established players such as David Warner, AB de Villiers, and Chris Gayle also feature among the top averages, reflecting both consistency and longevity in the IPL.

### Query 13: Batters with the Highest Strike Rate

#### Business Problem

Identify the batters who score at the fastest rate, considering only players who have faced a substantial number of legal deliveries.

```python
con.execute("""
SELECT
    batter,
    CAST(SUM(batsman_runs) AS INTEGER) AS total_runs,

CAST(
    SUM(
        CASE
            WHEN extras_type IS NULL
                 OR extras_type != 'wides'
            THEN 1
            ELSE 0
        END
    ) AS INTEGER 
) AS balls_faced,

    ROUND(
        100.0 * SUM(batsman_runs)
        / NULLIF(
            SUM(
                CASE
                    WHEN extras_type IS NULL
                         OR extras_type != 'wides'
                    THEN 1
                    ELSE 0
                END
            ),
            0
        ),
        2
    ) AS strike_rate

FROM deliveries_clean

GROUP BY batter

HAVING SUM(
    CASE
        WHEN extras_type IS NULL
             OR extras_type != 'wides'
        THEN 1
        ELSE 0
    END
) >= 500

ORDER BY strike_rate DESC, batter
LIMIT 10;
""").df()
```

### Observation 

Andre Russell has the highest strike rate (174.84) among batters who have faced at least 500 legal deliveries, followed by Heinrich Klaasen (168.31) and Sunil Narine (165.84). The list is dominated by aggressive middle-order batters and power hitters known for scoring rapidly rather than accumulating large totals.

### Query 17: Best Bowling Average

#### Business Problem

Identify the bowlers with the lowest bowling average among those who have taken at least 100 wickets.

```python
con.execute("""
WITH bowling_runs AS (

SELECT
    bowler,
    SUM(total_runs) AS runs_conceded
FROM deliveries_clean
GROUP BY bowler

),

bowling_wickets AS (

SELECT
    bowler,

    COUNT(*) AS wickets

FROM deliveries_clean

WHERE dismissal_kind IN (
    'bowled',
    'caught',
    'caught and bowled',
    'lbw',
    'stumped',
    'hit wicket'
)

GROUP BY bowler

)

SELECT

    br.bowler,

    CAST(br.runs_conceded AS INTEGER) AS runs_conceded,

    bw.wickets,

    ROUND(
        br.runs_conceded * 1.0 / bw.wickets,
        2
    ) AS bowling_average

FROM bowling_runs br

JOIN bowling_wickets bw
ON br.bowler = bw.bowler

WHERE bw.wickets >= 100

ORDER BY bowling_average ASC, br.bowler

LIMIT 10;
""").df()
```

### Observation 

Lasith Malinga has the best bowling average (20.51) among bowlers with at least 100 wickets, followed by Kagiso Rabada (22.10) and Rashid Khan (22.42). The rankings highlight bowlers who consistently took wickets while conceding relatively few runs over extended IPL careers.

### Query 18: Best Economy Rate

#### Business Problem

Identify the bowlers with the lowest economy rate among those who have bowled at least 300 overs.

```python
con.execute("""
WITH bowling_stats AS (

SELECT

    bowler,

    SUM(total_runs) AS runs_conceded,

    SUM(
        CASE
            WHEN extras_type IS DISTINCT FROM 'wides'
             AND extras_type IS DISTINCT FROM 'noballs'
            THEN 1
            ELSE 0
        END
    ) AS legal_balls

FROM deliveries_clean

GROUP BY bowler

)

SELECT

    bowler,

    CAST(runs_conceded AS INTEGER) AS runs_conceded,

    CAST(legal_balls AS INTEGER) AS balls,

    ROUND(
        runs_conceded * 6.0 / legal_balls,
        2
    ) AS economy

FROM bowling_stats

WHERE legal_balls >= 1800

ORDER BY economy ASC, bowler

LIMIT 10;
""").df()
```

### Observation 

Sunil Narine has the best economy rate (6.87 runs per over) among bowlers who have bowled at least 300 overs, followed by Rashid Khan (6.98) and Dale Steyn (7.10). The top rankings are dominated by spinners and disciplined pace bowlers who consistently restricted scoring over long IPL careers.

### Query 19: Best Bowling Strike Rate

#### Business Problem

Identify the bowlers who require the fewest deliveries to take a wicket, considering only bowlers with at least 100 wickets.

```python
con.execute("""
WITH bowling_balls AS (

SELECT

    bowler,

    SUM(
        CASE
            WHEN extras_type IS DISTINCT FROM 'wides'
             AND extras_type IS DISTINCT FROM 'noballs'
            THEN 1
            ELSE 0
        END
    ) AS legal_balls

FROM deliveries_clean

GROUP BY bowler

),

bowling_wickets AS (

SELECT

    bowler,

    COUNT(*) AS wickets

FROM deliveries_clean

WHERE dismissal_kind IN (
    'bowled',
    'caught',
    'caught and bowled',
    'lbw',
    'stumped',
    'hit wicket'
)

GROUP BY bowler

)

SELECT

    bb.bowler,

    CAST(bb.legal_balls AS INTEGER) AS balls,

    bw.wickets,

    ROUND(
        bb.legal_balls * 1.0 / bw.wickets,
        2
    ) AS strike_rate

FROM bowling_balls bb

JOIN bowling_wickets bw
ON bb.bowler = bw.bowler

WHERE bw.wickets >= 100

ORDER BY strike_rate ASC, bb.bowler

LIMIT 10;
""").df()
```

### Observation 

Andre Russell has the best bowling strike rate (14.76 balls per wicket) among bowlers with at least 100 wickets, followed by Kagiso Rabada (15.24) and Harshal Patel (16.01). These bowlers consistently take wickets at short intervals, highlighting their effectiveness as attacking wicket-takers in the IPL.

## Module 5: Advanced SQL Analytics

### Query 21: Orange Cap Winner by Season

#### Business Problem

Identify the highest run-scorer in every IPL season using a window-ranking function.

```python
con.execute("""
WITH season_runs AS (

    SELECT
        m.season,
        d.batter,
        SUM(d.batsman_runs) AS total_runs

    FROM deliveries_clean d

    JOIN matches_clean m
        ON d.match_id = m.id

    GROUP BY
        m.season,
        d.batter
),

ranked_batters AS (

    SELECT
        season,
        batter,
        total_runs,

        RANK() OVER (
            PARTITION BY season
            ORDER BY total_runs DESC
        ) AS run_rank

    FROM season_runs
)

SELECT
    season,
    batter,
    CAST(total_runs AS INTEGER) AS runs

FROM ranked_batters

WHERE run_rank = 1

ORDER BY season;
""").df()
```

### Observation 

Chris Gayle, Virat Kohli and David Warner are the only batters to win the Orange Cap multiple times, with Gayle topping the run charts in 2011 and 2012, Virat Kohli in 2016 and 2024 and Warner in 2015 and 2019. Virat Kohli's 973 runs in 2016 remains the highest run tally by an Orange Cap winner in a single IPL season.

### Query 22: Purple Cap Winner by Season

#### Business Problem

Identify the highest wicket-taker in every IPL season while counting only dismissals officially credited to the bowler.

```python
con.execute("""
WITH season_wickets AS (

    SELECT
        m.season,
        d.bowler,

        SUM(
            CASE
                WHEN d.dismissal_kind IN (
                    'bowled',
                    'caught',
                    'caught and bowled',
                    'lbw',
                    'stumped',
                    'hit wicket'
                )
                THEN 1
                ELSE 0
            END
        ) AS wickets

    FROM deliveries_clean d

    JOIN matches_clean m
        ON d.match_id = m.id

    GROUP BY
        m.season,
        d.bowler
),

ranked_bowlers AS (

    SELECT
        season,
        bowler,
        wickets,

        RANK() OVER (
            PARTITION BY season
            ORDER BY wickets DESC
        ) AS wicket_rank

    FROM season_wickets

    WHERE wickets > 0
)

SELECT
    season,
    bowler,
    CAST(wickets AS INTEGER) AS wickets

FROM ranked_bowlers

WHERE wicket_rank = 1

ORDER BY season;
""").df()
```

### Observation 

Dwayne Bravo, Bhuvneshwar Kumar, Harshal Patel and Kagiso Rabada are the only bowlers to finish as the highest wicket-taker in multiple IPL seasons. The 2019 season produced a tie for the Purple Cap, with Kagiso Rabada and Imran Tahir each taking 26 wickets, demonstrating why a ranking function is more appropriate than selecting only a single maximum.

### Query 23: Top Three Run-Scorers in Every Season

#### Business Problem

Rank batters within each season and identify the top three run-scorers for every IPL season.

```python
con.execute("""
WITH season_runs AS (

    SELECT
        m.season,
        d.batter,
        SUM(d.batsman_runs) AS total_runs

    FROM deliveries_clean d

    JOIN matches_clean m
        ON d.match_id = m.id

    GROUP BY
        m.season,
        d.batter
),

ranked_batters AS (

    SELECT
        season,
        batter,
        total_runs,

        DENSE_RANK() OVER (
            PARTITION BY season
            ORDER BY total_runs DESC
        ) AS season_rank

    FROM season_runs
)

SELECT
    season,
    season_rank,
    batter,
    CAST(total_runs AS INTEGER) AS runs

FROM ranked_batters

WHERE season_rank <= 3

ORDER BY
    season,
    season_rank,
    batter;
""").df()
```

### Observation 

The query ranks the top three run-scorers for every IPL season using DENSE_RANK(), preserving tied positions. ### Query 24: Running Team Score by Over

```python
con.execute("""
WITH over_scores AS (

    SELECT
        match_id,
        inning,
        batting_team,
        over,
        SUM(total_runs) AS runs_in_over

    FROM deliveries_clean

    GROUP BY
        match_id,
        inning,
        batting_team,
        over
),

score_progression AS (

    SELECT
        match_id,
        inning,
        batting_team,
        over,
        runs_in_over,

        SUM(runs_in_over) OVER (
            PARTITION BY match_id, inning
            ORDER BY over
            ROWS BETWEEN UNBOUNDED PRECEDING
                     AND CURRENT ROW
        ) AS cumulative_score

    FROM over_scores
)

SELECT
    match_id,
    inning,
    batting_team,
    over + 1 AS over_number,
    CAST(runs_in_over AS INTEGER) AS runs_in_over,
    CAST(cumulative_score AS INTEGER) AS cumulative_score

FROM score_progression

ORDER BY
    match_id,
    inning,
    over

LIMIT 100;
""").df()
```

### Observation 

The cumulative score increases over by over by adding each over’s runs to the previous total. For example, Kolkata Knight Riders scored 3 runs in the first over and reached 21 after the second over, showing that the SQL window function correctly tracks the running innings total.

### Query 25: Season-Wise Team Win Ranking

#### Business Problem

Calculate the matches played, matches won and win percentage of every team in each season, and rank teams based on their total wins.

```python
con.execute("""
WITH team_matches AS (

    SELECT
        season,
        team1 AS team
    FROM matches_clean

    UNION ALL

    SELECT
        season,
        team2 AS team
    FROM matches_clean
),

matches_played AS (

    SELECT
        season,
        team,
        COUNT(*) AS matches_played

    FROM team_matches

    GROUP BY
        season,
        team
),

matches_won AS (

    SELECT
        season,
        winner AS team,
        COUNT(*) AS matches_won

    FROM matches_clean

    WHERE winner IS NOT NULL

    GROUP BY
        season,
        winner
),

team_performance AS (

    SELECT
        mp.season,
        mp.team,
        mp.matches_played,
        COALESCE(mw.matches_won, 0) AS matches_won,

        ROUND(
            COALESCE(mw.matches_won, 0) * 100.0
            / NULLIF(mp.matches_played, 0),
            2
        ) AS win_percentage

    FROM matches_played mp

    LEFT JOIN matches_won mw
        ON mp.season = mw.season
       AND mp.team = mw.team
),

ranked_teams AS (

    SELECT
        season,
        team,
        matches_played,
        matches_won,
        win_percentage,

        DENSE_RANK() OVER (
            PARTITION BY season
            ORDER BY
                matches_won DESC,
                win_percentage DESC
        ) AS season_rank

    FROM team_performance
)

SELECT
    season,
    season_rank,
    team,
    CAST(matches_played AS INTEGER) AS matches_played,
    CAST(matches_won AS INTEGER) AS matches_won,
    win_percentage

FROM ranked_teams

ORDER BY
    season,
    season_rank,
    team;
""").df()
```

### Observation

Rajasthan Royals were the strongest team in the 2008 season, winning 13 of 16 matches with an 81.25% win rate. The use of DENSE_RANK() preserves tied positions when multiple teams record the same number of wins in a season.
